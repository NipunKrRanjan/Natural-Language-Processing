```python
df_html=pd.read_csv(r'C:\Users\KIIT0001\Desktop\NLP\Dataset\Spam Email raw text for NLP.csv')
print(df_html)

import re
def remove_html_tags(text):
    pattern=re.compile('<.*?>')
    return pattern.sub(r'',text)

df_html['MESSAGE']=df_html['MESSAGE'].apply(remove_html_tags)
df_html.to_csv(r'C:\Users\KIIT0001\Desktop\NLP\Dataset\Spam Email raw text for NLP.csv',index=True)
```

<h3> Removing Punctuations </h3>

```python
import string
exclude=string.punctuation
def remove_punct(text):
    return text.translate(str.maketrans('','',exclude))
df['text']=df['text'].apply(remove_punct)

print(df['text'])
```

<h3> Correcting Spellings </h3>

```python
import pandas as pd
import nltk
from textblob import TextBlob

# Download the corpus TextBlob needs for spell correction.
for corpus in ['words', 'wordnet', 'omw-1.4']:
    try:
        nltk.data.find(f'corpora/{corpus}' if corpus != 'words' else 'corpora/words')
    except LookupError:
        nltk.download(corpus, quiet=True)

# 1. Load dataset
path = r'C:\Users\KIIT0001\Desktop\NLP\Dataset\Incorrect Spelling.csv'
df_spell = pd.read_csv(path)
print(df_spell.head())

# 2. Ensure the correct column name is present
if 'augmented_text' in df_spell.columns:
    df_spell = df_spell.rename(columns={'augmented_text': 'incorrect_spelling'})
elif 'incorrect_spelling' not in df_spell.columns:
    raise ValueError("The CSV must contain either 'augmented_text' or 'incorrect_spelling'.")

# 3. Apply TextBlob spell correction safely

def correct_text(value):
    if pd.isna(value):
        return ''
    return str(TextBlob(str(value)).correct())


df_spell['corrected_spelling'] = df_spell['incorrect_spelling'].apply(correct_text)

# 4. Display the results
print(df_spell[['incorrect_spelling', 'corrected_spelling']].head())
```

<h3> Removing stopwords </h3> 

```python
from nltk.corpus import stopwords

def remove_stopwords(text):
    new_text=[]
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x=new_text[:]
    return " ".join(x)

df['removed_stopwords']=df['text'].apply(remove_stopwords,axis=1)
```
